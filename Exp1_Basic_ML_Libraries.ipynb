```python
# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
# Numpy Basics
# Creating NumPy Array

arr = np.array([10,20,30,40,50])

print("Array:")
print(arr)

print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Sum:", np.sum(arr))
```

```python
# Matrix Operations
matrix = np.array([[1,2],[3,4]])

print(matrix)

print("Transpose")
print(matrix.T)
```

```python
# Pandas Basics
data = {
    "Name":["A","B","C"],
    "Marks":[85,90,95]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())

print(df.info())

print(df.describe())
```

```python
# Matplotlib Basics
plt.plot(df["Name"],df["Marks"],marker='o')

plt.title("Student Marks")

plt.xlabel("Students")

plt.ylabel("Marks")

plt.show()
```

```python
# Load the dataset using scikit learn
iris = load_iris()

print(iris.feature_names)

print(iris.target_names)
```

```python
# Convert the dataset into dataframe
iris_df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

iris_df["Species"] = iris.target

iris_df.head()
```

```python
# Dataset Shape
print("Rows and Columns")

print(iris_df.shape)
```

```python
# Summary statistics
iris_df.describe()
```
