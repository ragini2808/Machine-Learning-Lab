```python
## 13.Outliers
## Method 1 Boxplot
plt.figure(figsize=(6, 3))
sns.boxplot(x=titanic["fare"], color="blue")
plt.title("Boxplot of Fare — Outliers appear as individual dots")
plt.show()
```

```python
## Method 2 IQR Rule
Q1 = titanic["fare"].quantile(0.25)
Q3 = titanic["fare"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = titanic[(titanic["fare"] < lower_bound) | (titanic["fare"] > upper_bound)]
print(f"Lower bound: {lower_bound:.2f}   Upper bound: {upper_bound:.2f}")
print("Number of outliers in 'fare':", outliers.shape[0])
```

```python
## Method 3  Z-Score
from scipy import stats

fare_no_na = titanic["fare"].dropna()
z_scores = np.abs(stats.zscore(fare_no_na))

z_outliers = fare_no_na[z_scores > 3]     # points more than 3 std-devs from the mean
print("Number of outliers using Z-score (|z| > 3):", z_outliers.shape[0])
```

## Outlier Detection for Penguins Dataset (using 'bill_length_mm')

### Method 1: IQR Rule for `bill_length_mm`

```python
Q1_bill = penguins["bill_length_mm"].quantile(0.25)
Q3_bill = penguins["bill_length_mm"].quantile(0.75)
IQR_bill = Q3_bill - Q1_bill

lower_bound_bill = Q1_bill - 1.5 * IQR_bill
upper_bound_bill = Q3_bill + 1.5 * IQR_bill

outliers_bill_iqr = penguins[(penguins["bill_length_mm"] < lower_bound_bill) | (penguins["bill_length_mm"] > upper_bound_bill)]
print(f"Lower bound for bill_length_mm: {lower_bound_bill:.2f}   Upper bound for bill_length_mm: {upper_bound_bill:.2f}")
print("Number of outliers in 'bill_length_mm' (IQR):", outliers_bill_iqr.shape[0])
```

### Method 2: Z-Score for `bill_length_mm`

```python
from scipy import stats

bill_length_no_na = penguins["bill_length_mm"].dropna()
z_scores_bill = np.abs(stats.zscore(bill_length_no_na))

z_outliers_bill = bill_length_no_na[z_scores_bill > 3] # points more than 3 std-devs from the mean
print("Number of outliers in 'bill_length_mm' (Z-score |z| > 3):", z_outliers_bill.shape[0])
```

### Visualization of Outliers for `bill_length_mm` in Penguins (Boxplot)

```python
plt.figure(figsize=(6, 3))
sns.boxplot(x=penguins["bill_length_mm"].dropna(), color="lightgreen")
plt.title("Boxplot of Bill Length (mm) — Outliers appear as individual dots")
plt.xlabel("Bill Length (mm)")
plt.show()
```

```python
## Removing Outliers (using IQR method bounds)
Q1_bill = penguins["bill_length_mm"].quantile(0.25)
Q3_bill = penguins["bill_length_mm"].quantile(0.75)
IQR_bill = Q3_bill - Q1_bill

lower_bound_bill = Q1_bill - 1.5 * IQR_bill
upper_bound_bill = Q3_bill + 1.5 * IQR_bill

penguins_no_outliers = penguins[(penguins["bill_length_mm"] >= lower_bound_bill) & (penguins["bill_length_mm"] <= upper_bound_bill)]
print("Original shape :", penguins.shape)
print("After removing bill_length_mm outliers:", penguins_no_outliers.shape)
```

```python
## 14.Feature Scaling

penguins_num = penguins.select_dtypes(include=np.number).dropna()
sample_cols = penguins_num.columns[:2]   # take 2 numeric columns to compare visually

scalers = {
    "Original":        penguins_num[sample_cols],
    "StandardScaler":  pd.DataFrame(StandardScaler().fit_transform(penguins_num[sample_cols]), columns=sample_cols),
    "MinMaxScaler":    pd.DataFrame(MinMaxScaler().fit_transform(penguins_num[sample_cols]),   columns=sample_cols),
    "RobustScaler":    pd.DataFrame(RobustScaler().fit_transform(penguins_num[sample_cols]),   columns=sample_cols),
    "Normalizer":      pd.DataFrame(Normalizer().fit_transform(penguins_num[sample_cols]),     columns=sample_cols),
}

fig, axes = plt.subplots(1, 5, figsize=(22, 4))
for ax, (name, data) in zip(axes, scalers.items()):
    sns.histplot(data[sample_cols[0]], ax=ax, kde=True, color="teal")
    ax.set_title(name)
plt.suptitle(f"Effect of Different Scalers on '{sample_cols[0]}' in Penguins Dataset")
plt.tight_layout()
plt.show()
```
